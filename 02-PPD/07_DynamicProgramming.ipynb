## 피보나치 수열

토끼가 달나리에 가면 다음과 같은 규칙으로 번식한다고 하자.
- 어른 토끼 암수 1쌍은 매달 아기 토끼 암수 1쌍씩 낳는다.  
- 아기 토끼는 태어난 지 1달이 지나면 어른 토끼가 되어 번식 가능해 진다.  
- 토끼는 늙지도 않고 죽지도 않는다.  

[질문1] 아기 토끼 암수 1쌍이 달나라에 정착하여 번식을 시작하면, 1년 뒤에 몇 쌍이 될까?  
[질문2] 5년(60개월) 후에 토끼 가족의 규모는 얼마나 될까?

토끼 가족의 규모는 피보나치 수열로 증가한다. 피보나치 수열은 다음과 같이 동작 한다.   
$
fib(n) =
\begin{cases}
fib(n-1) + fib(n-2)  & \text{if  } n > 1 \\
1  & \text{if  } n = 1 \\
0  & \text{if  } n = 0 \\
\end{cases}
$

- 피보나치 수열의 재귀호출 버전:

````python
def fib(n):
    if n > 1:
        return fib(n - 1) + fib(n - 2)
    else:
        return n

# Test code
fib(5)      # 5
````

- 피보나치 수열의 재귀 알고리즘 수행 시간을 계산해 보자

````python
def run_fib(n):
    from time import perf_counter
    start = perf_counter()
    answer = fib(n)
    finish = perf_counter()
    print("fib(", n, ") => ", answer, sep="")
    print(round(finish-start,4), "seconds")

# Test code
run_fib(12)
run_fib(24)
run_fib(36)
#run_fib(48)
#run_fib(60)
````

- 표채워풀기 알고리즘 버전:

````python
def fib(n):
    if n > 1:
        i = 2
        a, b = 0, 1
        while i <= n:
            a, b = b, a + b
            i += 1
        return b
    else:
        return n

# Test code
fib(5)      # 5
````

````python
def fibseq(n):
    fibs = [0, 1]
    for i in range(2, n+1):
        fibs.append(fibs[i-1] + fibs[i-2])
    return fibs

# Test code
fibseq(12)
````

- 파스칼의 삼각형 알고리즘 버전:

````python
def comb_pascal(n,r):
    row0 = [1 for _ in range(r+1)]      # 리스트 축약 표현
    matrix = [row0] + [[1] for _ in range(n-r)]
    for i in range(1, n-r+1):
        for j in range(1, r+1):
            newvalue = matrix[i][j-1] + matrix[i-1][j]
            matrix[i].append(newvalue)
    return matrix[n-r][r]

# Test code
comb(30, 3)     # 4060
````

- 파스칼의 삼각형 알고리즘 버전의 공간 절약:

    2차원 테이블을 구축할 필요 없이, 길이가 $r-1$ 인 리스트 하나만 있으면 충분하다. 왜 그럴까 다음의 코드를 살펴 보고 생각해 보자.

````python
def comb_pascal(n,r):
    vector = [1 for _ in range(r+1)]
    for _ in range(1, n-r+1):
        for j in range(1, r+1):
            vector[j] = vector[j-1] + vector[j]
    return vector[r]

# Test code
comb(30, 3)     # 4060
````

## 1까지 줄이는 최소 스텝

- 다음 3가지 규칙 중 하나를 선택하여 수행하는 과정을 반복 하면, 주어진 양수 $n$을 1로 줄일 수 있다.
  - 1을 뺸다.
  - 2로 나누어지면, 2로 나눈다.
  - 3으로 나누어지면, 3으로 나눈다.

- 임의 양수 $n$을 1까지 줄여가는 가장 짧은 경로의 길이를 구하는 `1까지 줄이는 최소 스텝 minimum steps to on` 문제를 푸는 재귀 표현은 다음과 같다.  
  
    $
    minsteps(n) =
    \begin{cases}
    1 + minimum \{ ministeps(n-1), minsteps(n/2), minsteps(n/3) \}   & \text{if  } n > 1  \\
    0 & \text{if  } n = 1
    \end{cases}
    $

- 재귀 함수 버전:

````python
def minsteps(n):
    if n > 1:
        steps = minsteps(n-1)
        if n % 2 == 0:
            steps = min(steps, minsteps(n//2))
        if n % 3 == 0:
            steps = min(steps, minsteps(n//3))
        return 1 + steps
    else:
        return 0
````

- 재귀 함수 버전의 실행 시간 측정

````python
def run_minsteps(n):
    from time import perf_counter
    start = perf_counter()
    answer = minsteps(n)
    finish = perf_counter()
    print("minsteps(", n, ") => ", answer, sep="")
    print(round(finish-start), "seconds")

# Test code
run_minsteps(10)    # 3
run_minsteps(23)    # 6
run_minsteps(237)   # 8
run_minsteps(317)   # 10
#run_minsteps(514)
#run_minsteps(711)
#run_minsteps(908)
````

- 메모해두기 알고리즘 버전:

````python
def minsteps(n):
    memo = [0 for _ in range(n+1)]
    def loop(n):
        if n > 1:
            if memo[n] == 0:
                steps = loop(n - 1)
                if n % 2 == 0:
                    steps = min(steps, loop(n // 2))
                if n % 3 == 0:
                    steps = min(steps, loop(n // 3))
                memo[n] = steps + 1
            return memo[n]
        else:
            return 0
    return loop(n)

# Test code
run_minsteps(23)    # 6
run_minsteps(237)   # 8
run_minsteps(317)   # 10
run_minsteps(514)
run_minsteps(711)
run_minsteps(908)
````

$n$개 원반을 출발말뚝에서 도착말뚝으로 옮기는 방법
1. (재귀) 출발말뚝에 있는 $n-1$개의 원만을 임시말뚝으로 옮긴다.
2. 출발말뚝의 맨 아래에 남아 있는 가장 큰 원반을 도착말뚝으로 옮긴다.
3. (재귀) 임시말뚝에 있는 $n-1$개의 원반을 도착말뚝으로 옮긴다.

![](https://en.wikipedia.org/wiki/File:Tower_of_Hanoi.jpeg)

- 재귀 호출 버전:

````python
def tower_of_hanoi(n, source, destin, temp):
    if n > 1:
        tower_of_hanoi(n-1, source, temp, destin)
        print("Move a disk from", source, "to", destin)
        tower_of_hanoi(n-1, temp, destin, source)
    else:
        print("Move a disk from", source, "to", destin)

# Test code
tower_of_hanoi(4, "A", "C", "B")
````

- 실행 시간 측정

    디스크 이동 횟수를 기록하기 위해서는 재귀 함수 바깥에 변수를 두어 이동 횟수를 세어 기록해야 하낟. 이를 위해서 전역(global)으로 `count` 변수를 추가 한다. 변경된 재귀 호출 버전과 실행 시간 측정 코드는 다음과 같다.  

````python
def tower_of_hanoi(n, source, destin, temp):
    global count
    if n > 1:
        tower_of_hanoi(n-1, source, temp, destin)
        count += 1
        tower_of_hanoi(n-1, temp, destin, source)
    else:
        count += 1

for n in [4,6,8,16,24,25,26,27,28]:
    count = 0
    from time import perf_counter
    start = perf_counter()
    tower_of_hanoi(n, "A", "C", "B")
    finish = perf_counter()
    cpu_time = round(finish-start,1)
    print(n, "disks:", count, "moves in", cpu_time, "seconds")
````

### 설탕 최적 배송

봉달은 설탕을 인터넷으로 주문받아 포장하여 배송하는 일을 관장하고 있다. 설탕은 봉지에 포장되어 있으며, 봉지는 2kg, 3kg, 5kg짜리의 세 종류가 있다. 주문은 2kg 이상, kg 단위로만 받는다. 회사의 배송 원칙은 봉지의 수가 최소가 되도록 포장하는 것이다.  
예를 들어 18kg을 주문받았다고 하자. 3kg 봉지 6개를 보장해도 18kg가 되지만, 5kg 봉지 3개와 3kg 봉지 1개를 포장하면 봉지 4개만으로 18kg을 포장할 수 있다.  
주문 양을 kg 단위로 받아서 최소 봉지 개수를 리턴하는 `minbags` 함수를 다음과 같이 재귀로 작성하였다.

````python
def minbags(n):
    if n > 1:
        if n == 2 or n == 3 or n == 5:
            return 1
        else:
            smallest = minbags(n-2)
            if n > 4:
                smallest = min(smallest, minbags(n-3))
            if n > 6:
                smallest = min(smallest, minbags(n-5))
        return 1 + smallest
    else:
        return 0
````

- 실행 시간 측정

    다음 테스트 함수로 실행해 보면, 인수의 크기가 증가함에 따라 실행시간이 기하급수적으로 증가하여 실용적으로 사용할 수 없다.

````python
def run_minbags(n):
    from time import perf_counter
    start = perf_counter()
    answer = minbags(n)
    finish = perf_counter()
    print("minbags(", n, ") => ", answer, sep="")
    print(round(finish-start,1), "seconds")


# # Test code
for i in range(36,51):
    run_minbags(i)
````
