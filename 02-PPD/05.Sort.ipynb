### 선택정렬

- 재귀 함수

````python
def selection_sort(xs):
    if xs != []:
        smallest = min(xs)
        xs.remove(smallest)
        return [smallest] + selection_sort(xs)
    else:
        return []

selection_sort([3, 5, 4, 1])
````

- 꼬리재귀 함수

````python
def selection_sort(xs):
    def loop(xs, ss):
        if xs != []:
            smallest = min(xs)
            xs.remove(smallest)
            return loop(xs, ss+[smallest])
        else:
            return ss
    return loop(xs, [])

selection_sort([3, 5, 4, 1])
````

`append`를 사용하여 다음과 같이 작성할 수도 있다.

````python
def selection_sort(xs):
    def loop(xs, ss):
        if xs != []:
            smallest = min(xs)
            xs.remove(smallest)
            ss.append(smallest)
            return loop(xs, ss)
        else:
            return ss
    return loop(xs, [])

selection_sort([3, 5, 4, 1])
````

- while 루프 함수

````python
def selection_sort(xs):
    ss = []
    while xs != []:
        smallest = min(xs)
        xs.remove(smallest)
        ss.append(smallest)
    return ss

selection_sort([3, 5, 4, 1])
````

### 합병정렬

정렬된 두 리스트를 합병하는 `merge` 함수가 있다고 가정하고, 합병정렬 알고리즘을 재귀 함수로 작성하면 다음과 같다.
```python
def merge_sort(xs):
    if len(xs) > 1:
        mid = len(xs) // 2
        return merge(merge_sort(xs[:mid]), merge_sort(xs[mid:]))
    else:
        return xs
```

- 합병 함수 merge 의 설계

  - 합병할 리스트 둘 다 정렬되어 있으므로, 전체에서 가장 작은 원소는 두 리스트의 선두원소 둘 중에서 작은 원소이다.
  - 작은 원소를 빼내어 기합병된 리스트 후미에 붙인다.
  - 이 합병 과정을 두 리스트 중에서 하나가 소진될 때까지 계속 반복한다.
  - 둘 중에 하나가 소진되면 남은 리스트를 이미 합병된 리스트 뒤에 붙이다.

- 정렬된 리스트 left와 right를 합병하는 재귀함수 merge의 반복조건 (합병을 지속할 조건)
  ```text
  left와 right 양쪽에 원소가 최소한 하나씩은 있음
  = left != [] and right != []
  = not (left == []) and not (right == [])
  = not (left ==[] or right == [])
  ```

````python
def merge(left, right):
    if not (left == [] or right == []):
        if left[0] <= right[0]:
            return [left[0]] + merge(left[1:], right)
        else:
            return [right[0]] + merge(left, right[1:])
    else:
        return left + right

# Test code
merge([18, 23, 32], [7, 11, 55, 99])    # [7, 11, 18, 23, 32, 55, 99]
````

- merge: 꼬리재귀 함수 버전

````python
def merge(left, right):
    def loop(left, right, ss):
        if not (left == [] or right == []):
            if left[0] <= right[0]:
                ss.append(left[0])
                left.pop(0)
            else:
                ss.append(right[0])
                right.pop(0)
            return loop(left, right, ss)
        else:
            return ss + left + right
    return loop(left, right, [])

# Test code
a = merge([18, 23, 32], [7, 11, 55, 99]) # [7, 11, 18, 23, 32, 55, 99]
print(a)
````

### 퀵정렬 Quicksort

영국의 유명한 컴퓨터과학자 C.A.R Hoare 가 1960년대에 고안한 정렬 알고리즘
- 정렬할 리스트에서 원소를 하나 임의로 선택한다. 이 원소는 리스틀 둘로 나눌 기준값이면 피봇 pivot 이라고 한다.
- 피봇보다 작은 원소는 피봇의 왼쪽에 모으고, 큰 원소는 피봇의 오론쪽에 모은다.
- 왼쪽 리스트와 오른쪽 리스트를 각각 재귀로 퀵 정렬 한다.
- 정렬한 왼쪽 리스트, 피봇, 정렬할 오른쪽 리스르틀 나란히 붙인다.

재귀 할수로 작성한 quicksort 함수는 다음과 같다.
```python
def quicksort(xs):
    if len(xs) > 1:
        pivot = xs[0]
        (left, right) = partition(pivot, xs[1:])
        return quicksoft(left) + [pivot] + quicksoft(right)
    else:
        return xs
```

- partition 함수

````python
def partition(pivot, xs):
    if xs != []:
        left, right = partition(pivot, xs[1:])
        if xs[0] <= pivot:
            left.append(xs[0])
        else:
            right.append(xs[0])
        return left, right
    else:
        return [], []

# Test code
partition(5, [7, 2, 1, 9, 4])   # ([4, 1, 2], [9, 7])
````

- partition: 꼬리재귀 함수 버전
  
  partition (재귀)함수를 다음 뼈대코드에 맞춰 꼬리재귀 함수로 변환 작성하시오.

````python
def partition(pivot, xs):
    def loop(xs, left, right):
        if xs != []:
            if xs[0] <= pivot:
                left = [xs[0]] + left
            else:
                right = [xs[0]] + right
            return loop(xs[1:], left, right)
        else:
            return left, right
    return loop(xs, [], [])

# Test code
print(partition(5, [7, 2, 1, 9, 4]))   # ([4, 1, 2], [9, 7])
````

- partition: while 루프 버전
  
  다음 뼈대코드를 바탕으로 while 루프 버전을 작성하시오.

````python
def partition(pivot, xs):
    left, right = [], []
    while xs != []:
        if xs[0] <= pivot:
            left.insert(0, xs.pop(0))
        else:
            right.insert(0, xs.pop(0))
    return left, right

# Test code
print(partition(5, [7, 2, 1, 9, 4]))   # ([4, 1, 2], [9, 7])
````

- partition: for 루프 버전
  
  다음 뼈대코드를 바탕으로 for 루프 버전을 작성하시오.

````python
def partition(pivot, xs):
    left, right = [], []
    for x in xs:
        if x <= pivot:
            left.insert(0, x)
        else:
            right.insert(0, x)
    return left, right
def quicksort(xs):
    if len(xs) > 1:
        pivot = xs[0]
        (left, right) = partition(pivot, xs[1:])
        return quicksort(left) + [pivot] + quicksort(right)
    else:
        return xs

# Test code
print(partition(5, [7, 2, 1, 9, 4]))   # ([4, 1, 2], [9, 7])
print(quicksort([6, 8, 1, 4, 2, 0, 5, 9, 3, 7]))
````
