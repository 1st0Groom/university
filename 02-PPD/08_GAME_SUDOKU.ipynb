### 4x4 스도쿠

[단계1] 무작위 스도쿠 정답보드 만들기

- 정답 보드 만들기

````python
import random
````

- 씨앗 정답보드 만들기

````python
def initialize_board_4x4():
    row0 = [1,2,3,4]
    random.shuffle(row0)
    row1 = row0[2:4] + row0[0:2]
    row2 = [row0[1], row0[0], row0[3], row0[2]]
    row3 = row2[2:4] + row2[0:2]
    return [row0, row1, row2, row3]
````

- 가로줄 바꾸기

````python
def shuffle_ribbons(board):
    top = board[:2]
    bottom = board[2:]
    random.shuffle(top)
    random.shuffle(bottom)
    return top + bottom
````

- 세로줄 바꾸기 --> 가로세로 바꾸기(transpose)

````python
def transpose(board):
    transposed = []
    size = len(board)
    for _ in range(size):
        transposed.append([])
    for row in board:
        for i in range(size):
            transposed[i].append(row[i])
    return transposed
````

- 게임에 제시될 정답보드를 생성하는 `create_solution_board_4x4` :

````python
def create_solution_board_4x4():
    board = initialize_board_4x4()
    board = shuffle_ribbons(board)
    board = transpose(board)
    board = shuffle_ribbons(board)
    board = transpose(board)
    return board

# Test code
print(create_solution_board_4x4())
````

````python
def get_integer(message,i,j):
    digit = input(message)
    while not (digit.isdigit() and i <= int(digit) <= j):
        digit = input(message)
    return int(digit)
````
