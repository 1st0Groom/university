## 자연수 수열의 합

### 재귀 함수

자연수의 귀납 정의를 활용하면, 기초 조건인 $n=0$의 경우 0까지 고민할 필요 없이 다음과 같다.

$ \text{sigma}(0) = 0 $ [base]

귀납에 해당하는 $n>0$의 경우 $n$까지의 합인 $\text{sigma}(n)$은 계산했다고 가정하고, $n+1$까지의 합인 $\text{sigma}$(n+1)은 $\text{sigma}(n)$에 $n+1$을 더하면 되므로 다음과 같이 식을 세울 수 있다.

$ \text{sigma}(n+1) = \text{sigma}(n) + (n+1) $ [induction]

두 등식을 통합하여 정리하면 다음과 같다.

$
\text{sigma} (n) =
\begin{cases}
\text{sigma} (n-1) + n  & \text{if } n>0 \\
0 & \text{if } n=0
\end{cases}
$

````python
def sigma(n):
    if n > 0:
        return sigma(n-1) + n
    else:
        return 0
````

````python
sigma(5)
````

### while 루프

````python
def sigma(n):
    n, total = n, 0
    while n > 0:
        n, total = n-1, n+total
    return total
````

함수를 호출할 때, 복귀(리턴)를 위한 정보를 메모리 기록한다. 일반 함수에서는 호출과 복귀가 반복되면서 사용되는 메모리도 할당과 해제가 반복된다. 하지만 재귀 함수는 종료 조건이 될 때까지 호출이 계속되고, 종료 조건 이후 복귀가 계속 된다. 호출이 계속되는 동안 메모리의 사용량도 비례하여 증가된다. 결국에는 함수 호출 정보를 담아 두기 위한 메모리가 모두 소진되어 실행 불능의 상태가 발생 한다.

- 프로그램 실행시에 할당한 스택 메모리 크기가 스택 오버플로우 발생에 영향을 준다.

## 거듭제곱

### 재귀 함수

````python
def power(b, n):
    if n > 0:
        return b * power(b, n-1)
    else:
        return 1

# Test code
power(2, 5)         # 32
````

### 꼬리재귀 함수

````python
def power(b, n):
    def loop(n, prod):
        if n > 0:
            return loop(n-1, b*prod)
        else:
            return prod
    return loop(n, 1)

# Test code
power(2, 5)         # 32
````

### while 루프

````python
def power(b, n):
    prod = 1
    while n > 0:
        prod = b * prod
        n = n - 1
    return prod

# Test code
power(2, 5)         # 32
````

### 나눠 풀기 (분할정복, divide-and-conquer) 알고리즘

**질문)**  power 함수는 재귀 버전을 꼬리재귀 및 while 루프 버전으로 변환하여 사용 공간을 절약할 수 있었으나, 계산시간은 모두 동일하다. 계산시간을 줄일 수는 없을까?

- 계산시간을 줄이는데 가장 흔한 방법은 나눠 풀기(분할정복) 이다. 문제 해결 대상을 나눠 각각 따로 푼 다음, 합치는 방법이다.
- 나눠 풀기 알고리즘을 거듭제곱 문제에 적용해 보자.

````python
# 재귀 함수
def power(b, n):
    if n > 0:
        if n % 2 == 0:
            return power(b * b, n // 2)
        else:
            return b * power(b, n - 1)
    else:
        return 1

# Test code
power(2, 5)         # 32
````

````python
# 꼬리재귀 함수
def power(b, n):
    def loop(b, n, prod):
        if n > 0:
            if n % 2 == 0:
                return loop(b * b, n // 2, prod)
            else:
                return loop(b, n - 1, b * prod)
        else:
            return prod
    return loop(b, n, 1)

# Test code
power(2, 5)         # 32
````

````python
# while 루프
def power(b, n):
    prod = 1
    while n > 0:
        if n % 2 == 0:
            b = b * b
            n = n // 2
        else:
            n = n - 1
            prod = b * prod
    return prod

# Test code
power(2, 5)         # 32
````

### 유클리드 알고리즘

$
\text{gcd}(m, n) =
\begin{cases}
\text{gcd}(n, m mod n) & \text{if } n \ne 0 \\
m                      & \text{if } n = 0
\end{cases}
$

````python
# 재귀 함수
def gcd(m, n):
    if n != 0:
        return gcd(n, m % n)
    else:
        return m

# Test code
print(gcd(48, 18))      # 6
print(gcd(18, 48))      # 6
print(gcd(192, 72))     # 24
print(gcd(18, 57))      # 3
print(gcd(0, 11))       # 11
````

### 나눠 풀기 알고리즘

````python
def even(n):
    return n % 2 == 0

def odd(n):
    return n % 2 == 1
````

````python
def gcd(m, n):
    if not (m == 0 or n == 0):
        if even(m) and even(n):
            return 2 * gcd(m // 2, n // 2)
        elif even(m) and odd(n):
            return gcd(m // 2, n)
        elif odd(m) and even(n):
            return gcd(m, n // 2)
        elif m <= n:
            return gcd(m, (n - m) // 2 )
        else:
            return gcd(n, (m - n) // 2 )
    else:
        if m == 0:
            return n
        else:
            return m


# Test code
print(gcd(48, 18))      # 6
print(gcd(18, 48))      # 6
print(gcd(192, 72))     # 24
print(gcd(18, 57))      # 3
print(gcd(0, 11))       # 11
````

곱셈 연산자가 없다면 다음과 같이 덧셈과 뺄셈 연산자를 이용한 재귀 방법으로 곱셈 함수를 구현할 수 있다.

$
m \times n =
\begin{cases}
m + m \times(n-1) & \text{if } n > 0 \\
0                 & \text{if } n = 0
\end{cases}
$

여기서 $m$과 $n$은 항상 자연수라고 가정한다. 즉, 음수 곱셈은 고려하지 않는다.

````python
# 재귀 함수
def mult(m, n):
    if n > 0:
        return m + mult(m, n - 1)
    else:
        return 0


# Test code
mult(3, 6)      # 18
````

````python
# 꼬리재귀 함수
def mult(m, n):
    def loop(n, ans):
        if n > 0:
            return loop(n - 1, m + ans)
        else:
            return ans
    return loop(n, 0)

# Test code
mult(3, 6)      # 18
````

### 러시아 농부 알고리즘

러시아 농부들은 두 수를 곱할 때 구구단 없이 덧셈, 두 배, 절반만 가지고 곱셈을 계산하는 방법을 사용했다. 곱셈 방법은 다음과 같다.
- 곱할 두 수를 나란히 적는다.
- 첫째 수는 두 배를 하고, 둘째 수는 절만을 하여(나머지는 버림) 다음 줄에 나란히 적는다.
- 이 과정을 둘째 수가 1이 될 떄까지 계속한다.
- 둘째 수가 짝수인 줄은 모두 지운다.
- 남은 줄의 첫째 수를 모두 더한 값이 답이다.

알고리즘을 재귀 함수로 구현하면 다음과 같다.

````python
def russian_mult(m,n):
    if n > 0:
        if n % 2 == 1:
            return m + russian_mult(2*m,n//2)
        else:
            return russian_mult(2*m,n//2)
    else: # n <= 0
        return 0

# Test code
russian_mult(57, 86)              # 4902
````

### 팩토리얼

$n$의 팩토리얼(factorial)은 1부터 $n$까지 모든 자연수의 곱으로 다음과 같이 정의한다.

$
n! = 1 \times 2 \times \cdots \times n = {\displaystyle \prod_{k=1}^{n} k }
$

- 재귀로 정의 하면 다음과 같다.

  $
n! =
\begin{cases}
(n-1)! \times n & \text{if } n > 1 \\
1               & \text{if } n = 1
\end{cases}
$

- 이것을 활용하여 일반 재귀 함수 fac를 작성하면 다음과 같다.

````python
def fac(n):
    if n > 1:
        return fac(n-1) * n
    else:
        return 1

# Test code
fac(10)         # 3628800
````

### 순열

자연수 $n$개 원소의 $k$ 순열(permutation)은 서로 다른 $n$개의 원소 중에서 $k$를 뽑아서 한 줄로 세우는 방법으로, 그 가능한 방법의 개수는 $P(n,k)$ 로 표현하며 다음 식으로 구할 수 있다.

$$
\begin{align*}
P(1, 1) &= 1 \\
P(2, 1) &= 2 \\
P(2, 2) &= 2 \times 1 = 2 \\
P(3, 1) &= 3\\
P(3, 2) &= 3 \times 2 = 6 \\
P(3, 3) &= 3 \times 2 \times 1 = 6 \\
P(4, 1) &= 4 \\
P(4, 2) &= 4 \times 3 = 12 \\
P(4, 3) &= 4 \times 3 \times 2 = 24 \\
P(4, 4) &= 4 \times 3 \times 2 \times 1 = 24 \\
\cdots \\
P(n, 1) &= n \\
P(n, k) &= n \times (n-1) \times (n-2) \times \cdots \times (n-k+1) \\
\end{align*}
$$
